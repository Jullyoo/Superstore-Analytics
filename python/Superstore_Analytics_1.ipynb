IMPORTANDO BIBLIOTECAS

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

LIMPEZA E VERIFICAÇÃO DOS DADOS

```python
# Lendo o arquivo CSV

df = pd.read_csv("data/Superstore.csv", encoding="latin1")
df.head()
```

```python
# Verificando informações do DataFrame

df.info()
```

```python
# Verificando tamanho do DataFrame (Linhas e Colunas)

df.shape
```

```python
# Verificando Colunas

df.columns
```

```python
# Descrevendo o DataFrame

df.describe()
```

```python
# Verificando valores nulos

df.isnull().sum()
```

```python
# Corrigindo datas

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
# Verificando informações novamente após a correção das datas

df.info()
```

```python
# Criando colunas de Ano, Mês e Dias de diferença entre a data de envio e a data do pedido (Shipping Days)

df['year'] = df['Order Date'].dt.year
df['month'] = df['Order Date'].dt.month

df['shipping_days'] = (df['Ship Date'] - df['Order Date']).dt.days
```

```python
# Gerando algumas KPIs para análise

total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_orders = df['Order ID'].nunique()
total_customers = df['Customer ID'].nunique()
total_products = df['Product ID'].nunique()

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Orders: {total_orders}")
print(f"Total Customers: {total_customers}")
print(f"Total Products: {total_products}")
```

ANALISES

```python
# Vendas por categoria

sales_category = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

sales_category.plot(kind='bar', figsize=(8, 5), color='skyblue')
plt.title('Vendas por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Vendas')
plt.show()
```

```python
# Lucro por categoria

profit_category = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)

profit_category.plot(kind='bar', figsize=(8, 5), color='lightgreen')
plt.title('Lucro por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Lucro')
plt.show()
```

```python
# Evolução das vendas ao longo dos anos

sales_year = df.groupby('year')['Sales'].sum()

sales_year.plot(kind='line', marker='o', figsize=(9, 5), color='orange')
plt.title('Evolução das Vendas ao Longo dos Anos')
plt.xlabel('Ano')
plt.ylabel('Vendas')
plt.show()
```

```python
# Evolução do lucro ao longo dos anos

profit_year = df.groupby('year')['Profit'].sum()

profit_year.plot(kind='line', marker='o', figsize=(9, 5), color='green')
plt.title('Evolução do Lucro ao Longo dos Anos')
plt.xlabel('Ano')
plt.ylabel('Lucro')
plt.show()
```

```python
# Top 10 produtos mais vendidos

top_products = df.groupby('Product Name')['Sales'].sum().nlargest(10)

top_products.plot(kind='barh', figsize=(8, 5), color='lightcoral')
plt.title('Top 10 Produtos Mais Vendidos')
plt.xlabel('Produto')
plt.ylabel('Vendas')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
# Produtos com maior prejuizo

bottom_products = df.groupby('Product Name')['Profit'].sum().nsmallest(10)

bottom_products.plot(kind='barh', figsize=(8, 5), color='salmon')
plt.title('Top 10 Produtos com Maior Prejuízo')
plt.xlabel('Prejuízo')
plt.ylabel('Produto')
plt.show()
```

```python
# Desconto x Lucro

sns.scatterplot(data=df, x='Discount', y='Profit', alpha=0.5)
plt.title('Desconto x Lucro')
plt.xlabel('Desconto')
plt.ylabel('Lucro')
plt.show()
```

```python
# Lucro por região

profit_region = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)

profit_region.plot(kind='bar', figsize=(8, 5), color='mediumpurple')
plt.title('Lucro por Região')
plt.xlabel('Região')
plt.ylabel('Lucro')
plt.show()
```
